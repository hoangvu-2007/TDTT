## W6A2
Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị
trí i), giữ nguyên thứ tự

```python
def tong_tich_luy(lst):
    tong_hien_tai = 0
    ket_qua = []
    for i in lst:
        tong_hien_tai += i
        ket_qua.append(tong_hien_tai)
    return ket_qua

lst = list(map(int,input().split()))
print(tong_tich_luy(lst))
```

##  W6A3
Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple
sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp
xếp.)

```python
def xoay_trai(tup, k):
    k = k % len(tup)
    tup1 = tup[:k]
    tup2 = tup[k:]
    tup_kq = tup2 + tup1
    return tup_kq

tup = tuple(map(int,input().split()))
k = int(input())
print(xoay_trai(tup, k))
```

## W6A4
Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp
lại, thêm vào list của key đó

```python
def anh_xa(chuoi):
    cac_cap = chuoi.split()
    dic = {}
    for cap in cac_cap:
        k, v = cap.split(":")
        if k not in dic:
            dic[k] = [v]
        else:
            dic[k].append(v)
    return dic

chuoi = input().strip()
print(anh_xa(chuoi))
```

## W6A9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của
key.

```python
def gop_dictionary(dic1, dic2):
    dic = dic1
    for k,v in dic2.items():
        if k not in dic:
            dic[k] = v
        else:
            dic[k] += v
    danh_sach_keys = list(dic.keys())
    n = len(danh_sach_keys)    
    for i in range(n):
        for j in range(0, n - i - 1):
            if danh_sach_keys[j] > danh_sach_keys[j + 1]:
                danh_sach_keys[j], danh_sach_keys[j + 1] = danh_sach_keys[j + 1], danh_sach_keys[j]
    dic_kq = {}
    for key in danh_sach_keys:
        dic_kq[key] = danh_sach_keys[key]    
    return dic_kq


dic1 = eval(input())
dic2 = eval(input())
print(gop_dictionary(dic1, dic2))
```

## W6A10
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra
list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple
có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của
số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên
trong tuple.

```python
def cap_tong(lst, k):
    lst_kq = []
    a = len(lst)
    for i in range(a):
        for j in range(i+1,a):
            if (lst[i] + lst[j] == k):
                tup = (lst[i],lst[j])
                if tup not in lst_kq:
                    lst_kq.append(tup)
    n = len(lst_kq)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if lst_kq[j][0] > lst_kq[j+1][0]:
                lst_kq[j], lst_kq[j+1] = lst_kq[j+1], lst_kq[j]
    return lst_kq
                
lst = list(map(int,input().split()))
k = int(input())
print(cap_tong(lst, k))
```

##  W6A11 [Tách số chẵn và lẻ từ tuple]
Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa
tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên
thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.

```python
def chan_le(tup):
    lst_le = []
    lst_chan = []
    for i in tup:
        if i%2 == 0:
            lst_chan.append(i)
        else:
            lst_le.append(i)
    tup_le = (lst_le)
    tup_chan = (lst_chan)
    return f"{tup_chan}\n{tup_le}"

tup = tuple(map(int,input().split()))
print(chan_le(tup))
```

## W6A12 [Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list.
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất
trong số đó.

```python
def xuat_hien_nhieu_nhat(lst):
    dic = {}
    xh_max = None
    so_lan_xh = 0
    for i in lst:
        if i not in dic:
            dic[i] = 1
        else:
            dic[i] += 1
    for k,v in dic.items():
        if v > so_lan_xh:
            so_lan_xh = v
            xh_max = k
        elif v == so_lan_xh:
            if k < xh_max:
                xh_max = k
    return xh_max
        
lst = list(map(int,input().split()))
print(xuat_hien_nhieu_nhat(lst))
```

## W6A13 [Đảo ngược key và value của dictionary]
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một
dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In
ra dictionary mới.

```python
def dao_nguoc_dict(lst):
    dic = {}
    for i in range(0, len(lst)-1, 2):
        dic[lst[i+1]] = lst[i]
    return dic

lst = list(input().split())
print(dao_nguoc_dict(lst))
```

## W6A15 [Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python
def loc_dict(lst, k):
    dic = {}
    dic_kq = {}
    for i in range(0, len(lst)-1, 2):
        dic[lst[i]] = lst[i+1]
    for key,value in dic.items():
        value = int(value)
        if value > k:
            dic_kq[key] = value

    return dic_kq

lst = list(input().split())
k = int(input())
print(loc_dict(lst, k))
```
